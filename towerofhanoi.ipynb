```python
def hanoi(n, source, temp, dest):
    if n <= 0:
        return

    # Move n-1 disks from source to temp using dest as helper
    hanoi(n - 1, source, dest, temp)

    # Move disk from source rod to destination rod
    disk = source['disks'].pop()
    dest['disks'].append(disk)
    print(f"Move disk {disk} from {source['name']} to {dest['name']}")

    # Check if destination rod holds all disks
    if len(dest['disks']) == total_disks:
        print("Tower of Hanoi completed!")

    # Move n-1 disks from temp to dest using source as helper
    hanoi(n - 1, temp, source, dest)

    # Creates list of disks in source rod based on the provided n value
def solve_hanoi(n, source, temp, dest):
    if n <= 0:
        print("Cannot solve  with current number of disks! Please try again.")
        return
        
    global total_disks
    total_disks = n

    source = {'name': source, 'disks': list(range(n, 0, -1))}
    temp = {'name': temp, 'disks': []}
    dest = {'name': dest, 'disks': []}

    hanoi(n, source, temp, dest)
```
